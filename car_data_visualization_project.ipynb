```python
import matplotlib.pyplot as plt
import pandas as pd

# EDA questions:
# • Does vehicle age reduce price?
# • Does mileage affect selling price?
# • Manual vs automatic — which is more expensive?
# • Which brands retain value better?
# • Which fuel type has the highest average price?

df=pd.read_csv(r"C:\Users\LENOVO\Downloads\cardekho_dataset.csv")
df
```

```python
#Does vehicle age reduce price?
plt.scatter(
    df['vehicle_age'],
    df['selling_price'],
    color="red",
    label="Vehicle Age and Selling Price"
)
plt.title("vehicle and price reduce")
plt.xlabel("vehicle_age ")
plt.ylabel("selling_price")
plt.legend()
plt.show()
```

```python
## Does mileage affect selling price?

plt.bar(
    df['mileage'],
    df['selling_price'],
    color='green',
    label='Vehicle'
)

plt.title('Mileage vs Selling Price')
plt.xlabel('Mileage')
plt.ylabel('Selling Price')

plt.legend()
plt.show()
```

```python
 # Manual vs automatic — which is more expensive
avg = df.groupby('transmission_type')['selling_price'].mean()

print(avg)
```

```python
# create bar plot
plt.bar(avg.index,
        avg.values,
        color=['red','green']
       )
# create title
plt.title(" Manual vs automatic Average  price")
plt.xlabel('transmission_type')
plt.ylabel('Average  price')
plt.show()
```

```python
# Which brands retain value better?
avg_price = df.groupby('brand')['selling_price'].mean()
print(avg_price)
```

```python
# create plot
plt.bar(
    avg_price.index,
    avg_price.values,
    color=['blue','yellow']
    
)
#create title
plt.title("brand value")
plt.xlabel("Average Selling Price")
plt.ylabel('brand')
# create legend
plt.legend()
plt.show()
```

```python
# Which fuel type has the highest average price
# calculate average
avg_fule_price=df.groupby('fuel_type')['selling_price'].mean()
avg_fule_price
```

```python
# create plot
plt.bar(
    avg_fule_price.index,
    avg_fule_price.values,
    color=['pink','blue']
    
)
# add title
plt.title('highest average price')
plt.xlabel('average price')
plt.ylabel('fuel_type')
plt.legend()
plt.show()
```
